```python
%matplotlib inline
import pylab as pl
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
def makeplots(regname, reg=None):
    for fn in glob.glob(f"*{regname}_only*.image*fits"):
        print(fn)
        cube = SpectralCube.read(fn)
        if reg is not None:
            spec = cube.subcube_from_regions(reg).mean(axis=(1,2))
        else:
            spec = cube.mean(axis=(1,2))
        pl.figure()
        spec.quicklook()
        pl.gca().set_title(fn)
```

```python
makeplots('Source1')
```

```python
import regions
from astropy import units as u
from astropy import coordinates
bn = regions.CircleSkyRegion(center=coordinates.SkyCoord('5:35:14.1058 -5:22:22.645', frame='icrs', unit=(u.hour,u.deg)),
                             radius=0.04*u.arcsec)
```
